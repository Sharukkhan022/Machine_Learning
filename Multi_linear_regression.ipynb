```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt


from sklearn.linear_model import LinearRegression
```

```python
data = {
    "Experience": [1, 2, 1, 2, 3, 3, 4, 4],
    "Skills":     [1, 1, 2, 2, 2, 3, 3, 4],
    "Salary":     [17, 22, 19, 24, 29, 31, 36, 38]
}

data = pd.DataFrame(data)
data
```

```python
X = data[["Experience", "Skills"]]
X
```

```python
y = data["Salary"]
y
```

```python
print(type(X))
print(type(y))
```

```python
model = LinearRegression()
```

```python
model.fit(X, y)
```

```python
print("Coefficients:", model.coef_)
```

```python
print("Intercept:", model.intercept_)
```

```python
y = model.intercept_ + model.coef_[0] * X["Experience"] + model.coef_[1] * X["Skills"]


new_person = [[5, 4]]

prediction = model.predict(new_person)

print("Predicted Salary:", prediction)
```

```python
y_pred = model.predict(X)

print(y_pred)
```

```python
residuals = y - y_pred

print(residuals)
```

```python
from sklearn.metrics import mean_squared_error

mse = mean_squared_error(y, y_pred)

print("MSE:", mse)
```

```python
import numpy as np

rmse = np.sqrt(mse)

print("RMSE:", rmse)
```

```python
from sklearn.metrics import r2_score

r2 = r2_score(y, y_pred)

print("R²:", r2)
```

```python
import numpy as np

experience_range = np.linspace(
    data["Experience"].min(),
    data["Experience"].max(),
    20
)

skills_range = np.linspace(
    data["Skills"].min(),
    data["Skills"].max(),
    20
)
```

```python
experience_grid, skills_grid = np.meshgrid(
    experience_range,
    skills_range
)
```

```python
salary_grid = (
    model.intercept_
    + model.coef_[0] * experience_grid
    + model.coef_[1] * skills_grid
)
fig = plt.figure()

ax = fig.add_subplot(111, projection="3d")

# Actual data points
ax.scatter(
    data["Experience"],
    data["Skills"],
    data["Salary"]
)

# Regression plane
ax.plot_surface(
    experience_grid,
    skills_grid,
    salary_grid,
    alpha=0.5
)

ax.set_xlabel("Experience")
ax.set_ylabel("Skills")
ax.set_zlabel("Salary")

ax.set_title("Multiple Linear Regression")

plt.show()
```
